```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
import os

fingers = ['Thumb', 'Index', 'Middle', 'Ring', 'Little']
dataset_name = 'BCIIV'
seed = 42
win_size = 2.8

decoder = 'HiLoFuseNet'
file_path1 = f'results/o5/varyingWindow/{dataset_name}_{decoder}_mse_seed{seed}_win{win_size}_o5_trajectory.npz'
trajData1 = np.load(file_path1, allow_pickle=True)

decoder = 'LSTM'
file_path2 = f'results/o5/varyingWindow/{dataset_name}_{decoder}_mse_seed{seed}_win{win_size}_o5_trajectory.npz'
trajData2 = np.load(file_path2, allow_pickle=True)

traj_true = trajData1['true'].item()
traj_pred1 = trajData1['pred'].item()
traj_pred2 = trajData2['pred'].item()

# select a subject and window
viz_win = [50, 950]
subject = 'sub0'

traj_true_1sub = np.array(traj_true[subject]).squeeze()
traj_pred1_1sub = np.array(traj_pred1[subject]).squeeze()
traj_pred2_1sub = np.array(traj_pred2[subject]).squeeze()

time = np.arange(traj_true_1sub.shape[0])
num_fingers = traj_true_1sub.shape[1]

cmap = plt.get_cmap('coolwarm')
color_hilo = cmap(11 / 12)
color_lstm = cmap(1 / 12)

cm = 1 / 2.54
fig, axes = plt.subplots(num_fingers, 1, figsize=(20 * cm, 5.5 * cm), sharex=True, dpi=300)

y_formatter = mticker.FormatStrFormatter('%.1f')
lines = []

for i in range(num_fingers):
    ax = axes[i]

    y_true = traj_true_1sub[viz_win[0]:viz_win[1], i]
    y_pred1 = traj_pred1_1sub[viz_win[0]:viz_win[1], i]
    y_pred2 = traj_pred2_1sub[viz_win[0]:viz_win[1], i]

    l_true = ax.plot(time[viz_win[0]:viz_win[1]], y_true, color='#bcbcbc', alpha=0.9, linewidth=1.2, label='Ground Truth')
    l_pred2 = ax.plot(time[viz_win[0]:viz_win[1]], y_pred2, color=color_lstm, alpha=0.9, linewidth=0.9, label=r"LSTM$_{\text{HGA}}$")
    l_pred1 = ax.plot(time[viz_win[0]:viz_win[1]], y_pred1, color=color_hilo, alpha=0.9, linewidth=0.9, label='HiLoFuseNet')

    if i == 0:
        lines = l_true + l_pred2 + l_pred1

    ax.yaxis.set_major_formatter(y_formatter)
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.spines['left'].set_visible(True)
    ax.spines['left'].set_color('black')
    ax.spines['left'].set_linewidth(0.6)

    if i < num_fingers - 1:
        ax.spines['bottom'].set_visible(False)
        ax.tick_params(axis='x', bottom=False, labelbottom=False)
        ax.tick_params(axis='y', which='both', left=True, labelleft=True, length=2.5, width=0.6, color='black', labelsize=6)
    else:
        ax.spines['bottom'].set_visible(True)
        ax.spines['bottom'].set_color('black')
        ax.spines['bottom'].set_linewidth(0.6)
        ax.tick_params(axis='both', which='both', left=True, bottom=True, labelleft=True, labelbottom=True,
                       length=2.5, width=0.6, color='black', labelsize=5.5)
        ax.set_xlabel('Time Step', fontsize=6, labelpad=4)

    ax.set_ylabel(fingers[i], fontsize=6, labelpad=3)
    ax.grid(False)

labels = [l.get_label() for l in lines]
fig.legend(lines, labels, loc='upper center', bbox_to_anchor=(0.53, 0.98),
           ncol=3, fontsize=6, frameon=False, columnspacing=1)

plt.subplots_adjust(top=0.91, bottom=0.14, left=0.15, right=0.96, hspace=0.25)

figPath = 'E:/my papers/finger decoding/HiLoFuseNet/figures/'
if os.path.exists(figPath):
    plt.savefig(figPath + 'subject1_bciiv.png', dpi=300)
plt.show()
```
