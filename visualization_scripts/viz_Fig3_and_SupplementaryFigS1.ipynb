Subject 3 from BCIIV dataset

```python
from pathlib import Path
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
import numpy as np

def tensor_to_numpy(value) -> np.ndarray:
    if isinstance(value, (list, tuple)):
        arrays = [tensor_to_numpy(item) for item in value]
        return arrays[0] if len(arrays) == 1 else np.stack(arrays)

    if hasattr(value, "detach"):
        return value.detach().cpu().numpy()

    return np.asarray(value)

FINGERS = ["Thumb", "Index", "Middle", "Ring", "Little"]

DATASET_NAME = "BCIIV"
SEED = 42
WIN_SIZE = 2.8
SUBJECT = "sub1"
VIZ_START, VIZ_END = 470, 1070

RESULTS_DIR = Path("results") / "o5" / "varyingWindow"

HILO_PATH = RESULTS_DIR / (
    f"{DATASET_NAME}_HiLoFuseNet_mse_seed{SEED}_"
    f"win{WIN_SIZE}_o5_trajectory.npz"
)

LSTM_PATH = RESULTS_DIR / (
    f"{DATASET_NAME}_LSTM_mse_seed{SEED}_"
    f"win{WIN_SIZE}_o5_trajectory.npz"
)

FIGURE_DIR = Path("figures")
FIGURE_DIR.mkdir(parents=True, exist_ok=True)


# load data
with np.load(HILO_PATH, allow_pickle=True) as hilo_data:
    true_by_subject = hilo_data["true"].item()
    hilo_by_subject = hilo_data["pred"].item()

with np.load(LSTM_PATH, allow_pickle=True) as lstm_data:
    lstm_true_by_subject = lstm_data["true"].item()
    lstm_by_subject = lstm_data["pred"].item()

ground_truth = tensor_to_numpy(
    true_by_subject[SUBJECT]
).squeeze()

hilo_pred = tensor_to_numpy(
    hilo_by_subject[SUBJECT]
).squeeze()

lstm_ground_truth = tensor_to_numpy(
    lstm_true_by_subject[SUBJECT]
).squeeze()

lstm_pred = tensor_to_numpy(
    lstm_by_subject[SUBJECT]
).squeeze()


# check the data
assert ground_truth.shape == hilo_pred.shape == lstm_pred.shape
assert ground_truth.ndim == 2
assert ground_truth.shape[1] == len(FINGERS)
assert 0 <= VIZ_START < VIZ_END <= ground_truth.shape[0]

assert np.allclose(
    ground_truth,
    lstm_ground_truth,
    equal_nan=True,
), "HiLoFuseNet and LSTM have different ground truth"

time_step = np.arange(VIZ_START, VIZ_END)

y_true = ground_truth[VIZ_START:VIZ_END]
y_lstm = lstm_pred[VIZ_START:VIZ_END]
y_hilo = hilo_pred[VIZ_START:VIZ_END]


# design the figure
CM = 1 / 2.54
MARK_TIME_STEP = None

cmap = plt.get_cmap('coolwarm')
color_hilo = cmap(11 / 12)
color_lstm = cmap(1 / 12)

COLORS = {
    "truth": "#A2A2A2",
    "lstm": color_lstm,
    "hilo": color_hilo,
}

plt.rcParams.update({
    "font.family": "DejaVu Sans",
    "font.size": 4,
    "axes.titlesize": 4,
    "axes.labelsize": 4,
    "xtick.labelsize": 4.5,
    "ytick.labelsize": 4.5,
    "legend.fontsize": 5,
    "axes.linewidth": 0.6,
    "lines.linewidth": 0.8,
})

fig, axes = plt.subplots(
    len(FINGERS),
    1,
    figsize=(10 * CM, 6 * CM),
    sharex=True,
    dpi=300,
)

for i, (ax, finger) in enumerate(zip(axes, FINGERS)):
    ax.plot(
        time_step,
        y_true[:, i],
        color=COLORS["truth"],
        lw=1.0,
        alpha=0.80,
        label="Ground truth",
        zorder=3,
    )

    ax.plot(
        time_step,
        y_lstm[:, i],
        color=COLORS["lstm"],
        lw=1.0,
        label=r"LSTM$_{\mathrm{HGA}}$",
        zorder=4,
    )

    ax.plot(
        time_step,
        y_hilo[:, i],
        color=COLORS["hilo"],
        lw=1.0,
        label="HiLoFuseNet",
        zorder=5,
    )

    if MARK_TIME_STEP is not None:
        ax.axvline(
            MARK_TIME_STEP,
            color="#7A7A7A",
            linewidth=0.6,
            linestyle=(0, (3, 2)),
            alpha=0.8,
            zorder=2,
        )

    ax.text(
        0.012,
        0.9,
        finger,
        transform=ax.transAxes,
        ha="left",
        va="top",
        fontsize=4.5,
        color="#303030",
        zorder=10,
    )

    ax.set_ylim(-1, 7)
    ax.set_yticks([0, 2, 4, 6])
    ax.set_xlim(VIZ_START, VIZ_END - 1)

    ax.set_axisbelow(True)
    ax.grid(False)

    # ax.spines[["top", "right"]].set_visible(False)
    ax.spines[["top", "right"]].set_color("#C3C3C3")
    ax.spines[["left", "bottom"]].set_zorder(0)
    ax.spines[["left", "bottom"]].set_color("#C3C3C3")
    ax.spines[["left", "bottom"]].set_linewidth(0.55)

    ax.tick_params(
        length=1.2,
        width=0.55,
        pad=2,
        color="#C3C3C3",
        labelcolor="#000000",
    )


axes[-1].set_xlabel(
    "Time step",
    fontsize=6,
    # loc="right",
    labelpad=4,
    color="#000000",
)

fig.supylabel(
    "Normalized Trajectory",
    x=0.07,
    fontsize=6,
    color="#000000",
)

legend_handles = [
    Line2D(
        [0], [0],
        color=COLORS["truth"],
        lw=1.0,
        alpha=0.80,
        label="Ground truth",
    ),
    Line2D(
        [0], [0],
        color=COLORS["lstm"],
        lw=1.0,
        label=r"LSTM$_{\mathrm{HGA}}$",
    ),
    Line2D(
        [0], [0],
        color=COLORS["hilo"],
        lw=1.0,
        label="HiLoFuseNet",
    ),
]

fig.legend(
    handles=legend_handles,
    loc="upper center",
    bbox_to_anchor=(0.5, 1.05),
    ncol=3,
    frameon=False,
    handlelength=2.6,
    columnspacing=1,
)

fig.subplots_adjust(
    left=0.12,
    right=0.985,
    bottom=0.12,
    top=0.95,
    hspace=0.25,
)

# save figure
output_path = FIGURE_DIR / (
    f"{DATASET_NAME}_{SUBJECT}_"
    f"t{VIZ_START}-{VIZ_END}_"
    f"HiLoFuseNet_vs_LSTM.png"
)
fig.savefig(
    output_path,
    dpi=300,
    bbox_inches="tight",
    facecolor="white",
)
plt.show()
```
